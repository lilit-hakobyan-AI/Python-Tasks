```python
import random

features = []
weights = []

for i in range(10):
  value = random.randint(0, 1)
  features.append(value)

for i in range(10):
  weight = random.randint(1, 10)
  weights.append(weight)

score = 0

for i in range(10):
  score += features[i] * weights[i]

threshold = sum(weights) * 0.5


print("Features:", features)
print("Weights:", weights)
print("score:", score)

if score >= 100:
  print("Approved")
else:
  print("Not approved")
```
