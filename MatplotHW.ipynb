```python
import numpy as np

stores = ["Yerevan", "Gyumri", "Vanadzor", "Abovyan", "Dilijan"]

laptops = np.array([120, 90, 75, 105, 65])
smartphones = np.array([250, 180, 160, 210, 140])
tablets = np.array([85, 60, 45, 70, 50])
accessories = np.array([310, 220, 190, 260, 175])

# 1. Total sales for each store
store_sales = laptops + smartphones + tablets + accessories
print("1. Total sales for each store:", store_sales)

# 2. Total sales for each product category
laptop_total = np.sum(laptops)
smartphone_total = np.sum(smartphones)
tablet_total = np.sum(tablets)
accessories_total = np.sum(accessories)

print("2. Total sales for each category:")
print("Laptops:", laptop_total)
print("Smartphones:", smartphone_total)
print("Tablets:", tablet_total)
print("Accessories:", accessories_total)

# 3. Highest-selling store
if store_sales[0] > store_sales[1] and  store_sales[0] > store_sales[2] and  store_sales[0] > store_sales[3] and  store_sales[0] > store_sales[4]:
    print("3. Highest_selling store: Yerevan")
elif store_sales[1] > store_sales[0] and  store_sales[1] > store_sales[2] and  store_sales[1] > store_sales[3] and  store_sales[1] > store_sales[4]:
    print("3. Highest_selling store: Gyumri")
elif store_sales[2] > store_sales[0] and  store_sales[2] > store_sales[1] and  store_sales[2] > store_sales[3] and  store_sales[2] > store_sales[4]:
    print("3. Highest_selling store: Vanadzor")
elif store_sales[3] > store_sales[0] and  store_sales[3] > store_sales[1] and  store_sales[3] > store_sales[2] and  store_sales[3] > store_sales[4]:
    print("3. Highest_selling store: Abovyan")
else:
  print("3. Highest-selling store: Dilijan")

# 4. Most popular product category

if laptop_total > smartphone_total and laptop_total > tablet_total and laptop_total > accessories_total:
  print("4. Most popular category: Laptops")

elif smartphone_total > laptop_total and smartphone_total > tablet_total and smartphone_total > accessories_total:
  print("4. Most popular category: Smartphones")

elif tablet_total > laptop_total and tablet_total > smartphone_total and tablet_total > accessories_total:
  print("4. Most popular category: Tablets")

else:
  print("4. Most popular category: Accessories")


# 5. Average laptop sales across stores
average_laptops = np.mean(laptops)
print("5. Average laptop sales:", average_laptops)

# 6. Percentage contribution of each store
total_sales = np.sum(store_sales)
percentages = store_sales / total_sales * 100

print("6. Percentage contribution of each store:")
print(percentages)

# 7. Difference between highest-selling and lowest-selling stores
difference = np.max(store_sales) - np.min(store_sales)
print("7. Difference:", difference)
```

```python
import numpy as np
import matplotlib.pyplot as plt

stores = np.array([
    "Yerevan", "Gyumri", "Vanadzor", "Abovyan", "Dilijan"
])

categories = np.array([
    "Laptops", "Smartphones", "Tablets", "Accessories"
])

sales = np.array([
    [120, 250, 85, 310],
    [90, 180, 60, 220],
    [75, 160, 45, 190],
    [105, 210, 70, 260],
    [65, 140, 50, 175]
])

# Create 5 separate graphs in one figure
fig, ax = plt.subplots(2, 3, figsize=(16, 9))

colors = ["royalblue", "orange", "green", "red"]

# Yerevan
ax[0, 0].bar(categories, sales[0], color=colors)
ax[0, 0].set_title("Yerevan")

# Gyumri
ax[0, 1].bar(categories, sales[1], color=colors)
ax[0, 1].set_title("Gyumri")

# Vanadzor
ax[0, 2].bar(categories, sales[2], color=colors)
ax[0, 2].set_title("Vanadzor")

# Abovyan
ax[1, 0].bar(categories, sales[3], color=colors)
ax[1, 0].set_title("Abovyan")

# Dilijan
ax[1, 1].bar(categories, sales[4], color=colors)
ax[1, 1].set_title("Dilijan")

# Hide the unused sixth graph
ax[1, 2].axis("off")

# Set same Y-axis scale for fair comparison
ax[0, 0].set_ylim(0, 350)
ax[0, 1].set_ylim(0, 350)
ax[0, 2].set_ylim(0, 350)
ax[1, 0].set_ylim(0, 350)
ax[1, 1].set_ylim(0, 350)

plt.suptitle("Electronics Sales by City", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

stores = ["Yerevan", "Gyumri", "Vanadzor", "Abovyan", "Dilijan"]

laptops = [120, 90, 75, 105, 65]
smartphones = [250, 180, 160, 210, 140]
tablets = [85, 60, 45, 70, 50]
accessories = [310, 220, 190, 260, 175]

plt.bar(stores, laptops, label="Laptops")
plt.bar(stores, smartphones, bottom=laptops, label="Smartphones")

bottom2 = np.array(laptops) + np.array(smartphones)
plt.bar(stores, tablets, bottom=bottom2, label="Tablets")

bottom3 = bottom2 + np.array(tablets)
plt.bar(stores, accessories, bottom=bottom3, label="Accessories")

plt.xlabel("Stores")
plt.ylabel("Total Sales")
plt.title("Total Sales by Store and Product Category")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

stores = ["Yerevan", "Gyumri", "Vanadzor", "Abovyan", "Dilijan"]

laptops = [120, 90, 75, 105, 65]
smartphones = [250, 180, 160, 210, 140]
tablets = [85, 60, 45, 70, 50]
accessories = [310, 220, 190, 260, 175]

laptop_total = np.array(laptops)
smartphone_total = np.array(smartphones)
tablet_total = np.array(tablets)
accessories_total = np.array(accessories)

store_sales = laptop_total + smartphone_total + tablet_total + accessories_total

total_sales = np.sum(store_sales)

percentages = store_sales / total_sales * 100

plt.pie(percentages, labels=stores, autopct="%1.1f%%")

plt.title("Store Sales Contribution")

plt.show()
```
