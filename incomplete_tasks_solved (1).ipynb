```python
# Problem 13

real_labels = [
    "spam", "not spam", "spam", "not spam",
    "spam", "not spam", "spam", "not spam"
]

predicted_labels = [
    "spam", "spam", "spam", "not spam",
    "not spam", "not spam", "spam", "spam"
]

tp = 0
tn = 0
fp = 0
fn = 0

for i in range(len(real_labels)):
    if real_labels[i] == "spam" and predicted_labels[i] == "spam":
        tp += 1
    elif real_labels[i] == "not spam" and predicted_labels[i] == "not spam":
        tn += 1
    elif real_labels[i] == "not spam" and predicted_labels[i] == "spam":
        fp += 1
    elif real_labels[i] == "spam" and predicted_labels[i] == "not spam":
        fn += 1

print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
```

```python
# Problem 14

instances = [
    [5, 12],
    [0, 1],
    [3, 8],
    [1, 2]
]

labels = ["spam", "not spam", "spam", "not spam"]

def dataset_summary(instances, labels):
    total_instances = len(instances)

    if total_instances > 0:
        features_per_instance = len(instances[0])
    else:
        features_per_instance = 0

    unique_labels = []

    for label in labels:
        if label not in unique_labels:
            unique_labels.append(label)

    print("Total instances:", total_instances)
    print("Features per instance:", features_per_instance)
    print("Unique labels:", unique_labels)

dataset_summary(instances, labels)
```

```python
# Problem 15

number_of_links = int(input("Number of links: "))
capital_letters = int(input("Number of capital letters: "))
answer = input("Does the email contain 'free'? yes/no: ")

contains_word_free = answer.strip().lower() == "yes"

conditions = 0

if number_of_links >= 4:
    conditions += 1

if capital_letters >= 10:
    conditions += 1

if contains_word_free:
    conditions += 1

if conditions >= 2:
    prediction = "spam"
else:
    prediction = "not spam"

print("Number of links:", number_of_links)
print("Capital letters:", capital_letters)
print("Contains 'free':", contains_word_free)
print("Prediction:", prediction)
```
