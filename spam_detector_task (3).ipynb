```python
features = []

weights = []

for i in range(20):
  value = int(input("Enter values for the features" + str(i + 1) + " (0 or 1): "))

  while  value != 0 and value != 1:
    value = int(input("Enter only 0 and 1"))
  features.append(value)

for i in range(20):
  weight = int(input("Enter weight for feature " + str(i + 1) + ": "))

  while weight <= 0:
    weight = int(input("Enter only numbers > 0"))
  weights.append(weight)

score = 0

for i in range(20):
  score = score + weights[i] * features[i]

print("Features", features)
print("Weights", weights)
print("Score", score)

if score >= 5:
  print("Spam")
else:
  print("Ham")

```
